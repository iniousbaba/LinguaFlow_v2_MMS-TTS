```python
# Cell 3 - Launch Flask + ngrok
import subprocess, sys, time, threading
from pyngrok import ngrok, conf

NGROK_AUTH_TOKEN = 'YOUR_NGROK_TOKEN_HERE'
APP_DIR = '/content/LinguaFlow_app'

conf.get_default().auth_token = NGROK_AUTH_TOKEN

flask_proc = subprocess.Popen(
    [sys.executable, 'app.py'],
    cwd=APP_DIR,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    bufsize=0,
)

time.sleep(3)

tunnel = ngrok.connect(5000)
print('LinguaFlow is live at:', tunnel.public_url)

# Stream Flask logs (runs until interrupted)
print('--- Flask logs (Ctrl+C or interrupt kernel to stop) ---')
try:
    for raw_line in iter(flask_proc.stdout.readline, b''):
        print(raw_line.decode('utf-8', errors='replace').rstrip())
except KeyboardInterrupt:
    pass
finally:
    print('--- Log stream stopped ---')
```
